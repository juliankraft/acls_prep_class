# Example notebook: loading a dataset and visualize it

This notebook shows a most basic workflow:

1. import libraries
2. load / download a dataset
3. take a first look at it
4. make a simple plot

We'll use the **penguins** dataset (measurements of three penguin species from Palmer Station, Antarctica), which ships as a built-in example dataset with `seaborn`.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## 1. Loading the dataset

`sns.load_dataset` downloads a CSV from the seaborn-data repository on GitHub the first time you call it, and caches it locally afterwards (so you'll need an internet connection at least once).

```python
penguins = sns.load_dataset("penguins")

print(f"shape: {penguins.shape}")
penguins.head()
```

## 2. Take a first look

Before plotting anything, it's good practice to check what you're working with: column types, missing values, basic summary statistics.

```python
penguins.info()
```

```python
penguins.describe()
```

## 3. Simple visualization

Let's plot bill length vs. bill depth, colored by species — a classic scatterplot for this dataset.

```python
sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
)
plt.title("Palmer penguins: bill length vs. bill depth")
plt.show()
```

And a quick look at body mass by species with a boxplot.

```python
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
)
plt.title("Body mass by species")
plt.show()
```
